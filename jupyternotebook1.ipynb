```python
def calc_total_bill(price, tax_rate = 0.13, tip_rate = 0.15):
    '''calc total bill with tax and tip'''
    tax = price *tax_rate
    tip = price * tip_rate
    total = tax + tip +price
    return total
```

```python
def eye_exam_covered(age, qualification_condition,time_since_last_exam):
    #assumption
    #age int
    #qualification_condition: boolean, True if you have an eligible medica condition False otherwise
    #time_since_last_exam: int, in months since last examination
    if 0 <= age <= 19:
        if time_since_last_exam>=12:
            return "you are eligible for one major eye exam or any minor assessment needed"
        else:
            return "It hasn't been 12 months since your last exam then you are not eligible for major eye exam but minor assessment."
        
    elif 20 <= age <=64:
        if qualification_condition:
            if time_since_last_exam>=12:
                return "you are eligible for one major eye exam or any minor assessment needed"
            else:
                return " It hasn't been 12 months since your last exam then you are not eligible for major eye exam but minor assessment."
        else:
            return " you do not have eligible medical condition"
    elif age >= 65:
        if time_since_last_exam >= 18:
            return " you are eligible for one major eye exam or 2 follow-up minor assessment needed"
        else: 
            return " It hasn't been 18 months since your last exam but you have 2 minor assessment available."
    else: 
        return "Invalid age"
    
```

```python
eye_exam_covered(-5,False, 14)
```

```python
eye_exam_covered(15, True, 3)
```

```python
eye_exam_covered(50, False, 5)
```
